#### W6A16:
Nhập vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải

```python
m,n = map(int,input().split())
a = []
for i in range(n):
    b = list(map(int,input().split()))
    if len(b) != m:
        print("Errol")
        break
    a.append(b)
for i in a:
    for k in i:
        print(f"{k:>4}",end="")
    print()

```

#### W6A17:
Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và đường chéo phụ của mảng

```python
n = int(input())
a = []
for i in range(n):
    b = list(map(int,input().split()))
    a.append(b)
duong_cheo_chinh = []
duong_cheo_phu = []
for i in range(len(a)):
    duong_cheo_chinh.append(a[i][i])
    duong_cheo_phu.append(a[i][n-1-i])
print(duong_cheo_chinh,duong_cheo_phu,sep = "\n")
```

#### W6A20:
Viết hàm đưa vào một list số nguyên và một số nguyên k. Hãy tìm và trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về -1

```python
def tim_k(a):
    k = int(input())
    for i in range(len(a)):
        if a[i] == k:
            return i+1
    else:
        return -1
a = list(map(int,input().split()))
print(tim_k(a))
```

#### W6A24:
Viết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 0 2 3 2 4 9

```python
def push_zeros(a):
    c = []
    d = []
    for i in a:
        if i == 0:
            c.append(i)
        else:
            d.append(i)
    return c+d

a = list(map(int,input().split()))
print(push_zeros(a))
```

#### W6A26:
Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2 phần tử đầu và cuối

```python
def cuc_tri(a):
    tong = 0
    for i in range(1,len(a)-1):
        if a[i-1] < a[i] > a[i+1]:
            tong += a[i]
    return tong
a = list(map(int,input().split()))
print(cuc_tri(a))
```

#### W6A27:
Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước), nếu không có in ra -1

```python
a = list(map(int,input().split()))
x = int(input())
b = []
for i in range(len(a)):
    if a[i] < x:
        b.append(i)
if not b:
    print(-1)
else:
    print(b)
```
