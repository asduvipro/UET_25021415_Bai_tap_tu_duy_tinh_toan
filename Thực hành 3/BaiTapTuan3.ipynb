 Viết một chương trình nhập vào một số nguyên n và in ra các chữ số đảo ngược nhau. 

```python
# Cách 1: biến về dạng chuỗi kí tự và dùng đảo ngược chuỗi
a = input()
a = str(a)
print("".join(reversed(a)))
    
```

```python
# cách 2: dùng phương pháp cắt lát
a = int(input())
print(str(a)[::-1])
```

```python
# Dùng phương pháp sử dụng vòng lặp
a = int(input())
b = len(str(a))
for i in range(b):
    print(a%10,end="")
    a = a//10
```

#### W3A3:
Kiểm tra xem một số có phải là lũy thừa của 2 hay không
Sử dụng các toán tử bitwise để xác định xem một số có phải là lũy thừa của 2 hay không.
Gợi ý: Đối với lũy thừa của 2, n & (n-1) == 0

```python
a = int(input())
if a & (a-a) == 0:
    print("a là một lũy thừa của 2")
```
