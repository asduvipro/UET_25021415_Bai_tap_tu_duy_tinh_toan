### Problem 1

**Task:** Write a function that counts the number of characters in a given string `s` and returns that count.  
If `s` is not a string, convert it to a string and count its characters.

```python
def count_chars(s):
    try:
        total = 0
        for _ in s:
            total += 1
        return total
    except TypeError:
        t = str(s)
        total = 0
        for _ in t:
            total += 1
        return total
```

```python
assert count_chars("") == 0
assert count_chars("a") == 1
assert count_chars("hello") == 5
assert count_chars("  spaces ") == 9
assert count_chars(12345) == 5
assert count_chars(-12345) == 6
assert count_chars(-12345.00) == 8
```

### Problem 2

**Task:** Write a function that counts the number of words in a string.  
A word is a maximal sequence of non-space characters. Multiple spaces and leading/trailing spaces may appear.  
If input is not a string, convert it to a string first.

```python
def count_words(s):
    try:
        text = s
    except TypeError:
        text = str(s)
    if not isinstance(text, str):
        text = str(text)
    count = 0
    in_word = False
    for ch in text:
        if ch != ' ' and ch != '\t' and ch != '\n' and ch != '\r' and ch != '\f' and ch != '\v':
            if not in_word:
                count += 1
                in_word = True
        else:
            in_word = False
    return count
```

```python
assert count_words("") == 0
assert count_words("one") == 1
assert count_words("two  words") == 2
assert count_words("  leading and trailing  ") == 3
assert count_words("a\tb\nc") == 3
assert count_words(1245) == 1
assert count_words(1245.00) == 1
```

### Problem 3

**Task:** Write a function that returns `True` if the input string contains **all 26 English letters** (a–z) at least once, ignoring case; otherwise return `False`.  
Non-letter characters are ignored.

```python
def is_pangram(s):
    try:
        text = s
        missing = [True]*26
        for ch in text:
            c = ch
            if c >= 'A' and c <= 'Z':
                c = chr(ord(c) + 32)
            if c >= 'a' and c <= 'z':
                idx = ord(c) - ord('a')
                if missing[idx]:
                    missing[idx] = False
        i = 0
        while i < 26:
            if missing[i]:
                return False
            i += 1
        return True
    except Exception:
        return False
```

```python
assert is_pangram("The quick brown fox jumps over a lazy dog") is True
assert is_pangram("Help") is False
assert is_pangram("") is False
assert is_pangram("Sphinx of black quartz, judge my vow!") is True
assert is_pangram("ABCDEFGHIJKLMNOPQRSTUVWXYZ") is True
assert is_pangram(1245) is False
```

### Problem 4

**Task:** Write a function that checks whether a given string is a palindrome (reads the same forward and backward).  
The check is **exact** (case and spaces matter). Return `True` or `False`. If input is not a string, convert it to a string.

```python
def is_palindrome(s):
    try:
        text = s
    except Exception:
        text = str(s)
    if not isinstance(text, str):
        text = str(text)
    i = 0
    j = 0
    for _ in text:
        j += 1
    j -= 1
    while i < j:
        if text[i] != text[j]:
            return False
        i += 1
        j -= 1
    return True
```

```python
assert is_palindrome("") is True
assert is_palindrome("a") is True
assert is_palindrome("abba") is True
assert is_palindrome("abc") is False
assert is_palindrome("Aba") is False
assert is_palindrome(12321) is True
assert is_palindrome(123212) is False
```

### Problem 5

**Task:** Given a full name string with Last, Middle, First (e.g., `"Nguyen Van A"`), return the reordered name as `First Middle Last` (e.g., `"A Van Nguyen"`).  
Assume tokens are separated by spaces; multiple spaces may exist. If there is only one token, return it unchanged.

```python
def reorder_name(fullname):
    if not isinstance(fullname, str):
        s = str(fullname)
    parts = []
    word = ""
    for ch in s:
        if ch != ' ':
            word += ch
        else:
            if word != "":
                parts.append(word)
                word = ""
    if word != "":
        parts.append(word)
    n = 0
    for _ in parts:
        n += 1
    result_tokens = []
    result_tokens.append(parts[n-1])
    i = 1
    while i < n-1:
        result_tokens.append(parts[i])
        i += 1
    result_tokens.append(parts[0])
    out = ""
    k = 0
    m = 0
    for _ in result_tokens:
        m += 1
    while k < m:
        out += result_tokens[k]
        if k != m-1:
            out += " "
        k += 1
    return out
```

```python
assert reorder_name("Nguyễn Văn A") == "A Văn Nguyễn"
assert reorder_name("Nguyen Van A") == "A Van Nguyen"
assert reorder_name("  Le   Thi   B   ") == "B Thi Le"
assert reorder_name("Vien A I") == "I A Vien"
```

### Problem 6

**Task:** Given an integer `n`, return the sum of its digits. Negative signs are ignored (use absolute value).

```python
def sum_of_digits(n):
    x = int(n)
    if x < 0:
        x = -x
    s = 0
    if x == 0:
        return 0
    while x > 0:
        s += x % 10
        x //= 10
    return s
```

```python
assert sum_of_digits(0) == 0
assert sum_of_digits(12345) == 15
assert sum_of_digits(-907) == 16
assert sum_of_digits(111) == 3
assert sum_of_digits(-5) == 5
```

### Problem 7

**Task:** Given a non-negative integer `n`, form the **largest** and **smallest** numbers that can be made by reordering its digits.  
Return them as a list `[max_value, min_value]` (avoiding tuples).  
For the smallest value, do not allow a leading zero (e.g., `2015 → 5210` and `1025`).

```python
def extremes_from_digits(n):
    x = int(n)
    if x < 0:
        x = -x
    counts = [0]*10
    if x == 0:
        counts[0] = 1
    else:
        while x > 0:
            d = x % 10
            counts[d] += 1
            x //= 10
    max_str = ""
    d = 9
    while d >= 0:
        c = counts[d]
        i = 0
        while i < c:
            max_str += chr(ord('0') + d)
            i += 1
        d -= 1
    min_str = ""
    first = -1
    d = 1
    while d <= 9 and first == -1:
        if counts[d] > 0:
            first = d
        d += 1
    if first == -1:
        min_str = "0"
    else:
        min_str += chr(ord('0') + first)
        counts[first] -= 1
        i = 0
        while i < counts[0]:
            min_str += '0'
            i += 1
        d = 1
        while d <= 9:
            i = 0
            while i < counts[d]:
                min_str += chr(ord('0') + d)
                i += 1
            d += 1
    max_val = 0
    for ch in max_str:
        max_val = max_val*10 + (ord(ch)-48)
    min_val = 0
    for ch in min_str:
        min_val = min_val*10 + (ord(ch)-48)
    return [max_val, min_val]
```

```python
assert extremes_from_digits(2015) == [5210, 1025]
assert extremes_from_digits(0) == [0, 0]
assert extremes_from_digits(1000) == [1000, 1000]
assert extremes_from_digits(1002) == [2100, 1002]
assert extremes_from_digits(5550) == [5550, 5055]
assert extremes_from_digits(1001) == [1100, 1001]
assert extremes_from_digits(90000000000000000) == [90000000000000000, 90000000000000000]
```

### Problem 8

**Task:** Given a lowercase string, return the count of the character `'a'`.  
If input is not a string, convert it first.

```python
def count_a(s):
    text = s
    cnt = 0
    for ch in text:
        if ch == 'a':
            cnt += 1
    return cnt
```

```python
assert count_a("") == 0
assert count_a("a") == 1
assert count_a("banana") == 3
assert count_a("abcdea") == 2
```

### Problem 9

**Task:** Input an integer `n`, count how many of its digits are **prime digits**  (2, 3, 5, 7).  
Negative signs are ignored.

```python
def count_prime_digits(n):
    x = int(n)
    if x < 0:
        x = -x
    if x == 0:
        return 0
    cnt = 0
    while x > 0:
        d = x % 10
        if d == 2 or d == 3 or d == 5 or d == 7:
            cnt += 1
        x //= 10
    return cnt
```

```python
assert count_prime_digits(1234) == 2
assert count_prime_digits(0) == 0
assert count_prime_digits(-27553) == 5
assert count_prime_digits(4680) == 0
```

### Problem 10

**Task:** Given a time string in the form `"hh:mm:ss"`, return the total number of seconds.  
Hours, minutes, and seconds must be integers with `0 ≤ mm < 60` and `0 ≤ ss < 60`.  
Return `-1` if values are out of range.

```python
def hms_to_seconds(s):
    try:
        text = s
        parts = []
        cur = ""
        for ch in text:
            if ch == ':':
                parts.append(cur)
                cur = ""
            else:
                cur += ch
        parts.append(cur)
        c = 0
        for _ in parts:
            c += 1
        if c != 3:
            return -1
        h = int(parts[0])
        m = int(parts[1])
        sec = int(parts[2])
        if m < 0 or m >= 60 or sec < 0 or sec >= 60 or h < 0:
            return -1
        return h*3600 + m*60 + sec
    except Exception:
        return -1
```

```python
assert hms_to_seconds("01:06:40") == 4000
assert hms_to_seconds("00:00:00") == 0
assert hms_to_seconds("10:59:59") == 39599
assert hms_to_seconds("1:6:40") == 4000
assert hms_to_seconds("23:60:00") == -1
```

### Problem 11

**Task:** Implement Atbash decoding for English letters. Map `a→z, b→y, ..., z→a` and similarly `A→Z, ..., Z→A`.  
Non-letters remain unchanged.

```python
def atbash_decode(x):
    try:
        s = x
    except Exception:
        s = str(x)
    if not isinstance(s, str):
        s = str(s)
    out = ""
    for ch in s:
        c = ch
        if c >= 'a' and c <= 'z':
            mirrored = 97 + (122 - ord(c))
            out += chr(mirrored)
        elif c >= 'A' and c <= 'Z':
            mirrored = 65 + (90 - ord(c))
            out += chr(mirrored)
        else:
            out += c
    return out
```

```python
assert atbash_decode("abcdefghijklmnopqrstuvwxyz") == "zyxwvutsrqponmlkjihgfedcba"
assert atbash_decode("Zyx") == "Abc"
assert atbash_decode("Hello, World!") == "Svool, Dliow!"
assert atbash_decode("") == ""
assert atbash_decode(123) == "123"
```

### Problem 12

**Task:** A string is **perfect** if it contains at least one uppercase letter, at least one lowercase letter, and at least one digit.  
Return `True` if the string is perfect; otherwise return `False`. If input is not a string, convert it.

```python
def is_perfect_string(s):
    try:
        text = s
    except Exception:
        text = str(s)
    if not isinstance(text, str):
        text = str(text)
    has_upper = False
    has_lower = False
    has_digit = False
    for ch in text:
        if ch >= 'A' and ch <= 'Z':
            has_upper = True
        elif ch >= 'a' and ch <= 'z':
            has_lower = True
        elif ch >= '0' and ch <= '9':
            has_digit = True
    return has_upper and has_lower and has_digit
```

```python
assert is_perfect_string("Aa0") is True
assert is_perfect_string("abc") is False
assert is_perfect_string("ABC") is False
assert is_perfect_string("123") is False
assert is_perfect_string("Passw0rd") is True
```
