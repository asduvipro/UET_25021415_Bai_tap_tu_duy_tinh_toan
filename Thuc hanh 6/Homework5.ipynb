#### W5A16:
Viết hàm kiểm tra một số có phải số Amstronhg hay không? Hàm trả về kiểu bool, có tham số là một số
nguyên x

```python
def amstronhg_number(x)->(int):
    a = len(str(x))
    c,d = 0,x
    for i in range(len(str(x))):
        c += (x%10)**a
        x //= 10
    return c == d
x = int(input())
print(amstronhg_number(x))
```

#### W5A17:
Viết hàm kiểm tra một số có phải số chính phương hay không? Hàm trả về kiểu bool, có tham số là một số nguyên x

```python
def so_chinh_phuong(x):
    for i in range(1,int(x**0.5)+1):
        if i**2 == x:
            return True
    else:
        return False
x = int(input())
print(so_chinh_phuong(x))
```

```python
def is_magic_number(x):
    c = 0
    while True:
        c += x%10
        x //= 10        
        if x == 0:
            x = c
            if len(str(c)) == 1:
                break
            c = 0
    return c == 1
x = int(input())
print(is_magic_number(x))


```

#### W5A20:
Viết hàm detect_number_type(s) trả về 'int' nếu s là số nguyên, 'float' nếu là số thực, ngược lại 'neither'.

```python
def detect_number_type(s):
    try:
        int(s)
        return "int"
    except ValueError:
        try:
            float(s)
            return "float"
        except ValueError:
            return "string"
s = input()
print(detect_number_type(s))
```

#### W5A23:
Viết hàm quadratic_roots_count(a, b, c) trả về số nghiệm thực của phương trình a*x^2 + b*x + c = 0.

```python
def quadratic_roots_count(a, b, c):
    d = b**2-4*a*c
    if d > 0:
        return ((-b-d**0.5)/(2*a),(-b+d**0.5)/(2*a))
    elif d == 0:
        return -b/(2*a)
    else:
        return "No real roots"
a, b, c = map(float, input().split())
print(quadratic_roots_count(a, b, c))
```

#### W5A24:
Viết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không.

```python
def are_collinear(x1,y1,x2,y2,x3,y3):
    return (x2-x1)/(x3-x1) == (y2-y1)/(y3-y1)
x1, y1, x2, y2, x3, y3 = map(float, input().split())
print(are_collinear(x1,y1,x2,y2,x3,y3))
```

#### W5A25:
Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay không. Trong đó: quad là một tứ giác (4 đỉnh = 4 điểm), p là một điểm cần kiểm tra. VD: quad = [(0,0), (4,0), (4,3), (0,3)], p = (2,1)

```python
def point_in_quadrilateral(quad, p)->(list, tuple):
    def cross_product(a,b,c):
        return (b[0]-a[0])*(c[1]-a[1])-(b[1]-a[1])*(c[0]-a[0])
    d = []
    for i in range(4):
        d.append(cross_product(quad[i],quad[(i+1)%4],p))
    if d[0] > 0 and d[1] > 0 and d[2] > 0 and d[3] > 0:
        return True
    elif d[0] < 0 and d[1] < 0 and d[2] < 0 and d[3] < 0:
        return True
    else:
        return False
quad = [tuple(map(float, input().split())) for _ in range(4)]
p = tuple(map(float, input().split()))
print(point_in_quadrilateral(quad, p))
```

#### W5A28:
Viết hàm tính góc ∠ABC (độ) từ toạ độ A(x1,y1), B(x2,y2), C(x3,y3).

```python
def angle(a,b,c)->tuple:
    import math
    vba = (a[0]-b[0],a[1]-b[1])
    vbc = (c[0]-b[0],c[1]-b[1])
    dot_product = vba[0]*vbc[0] + vba[1]*vbc[1]
    ba = math.sqrt(vba[0]**2 + vba[1]**2)
    bc = math.sqrt(vbc[0]**2 + vbc[1]**2)
    cos = dot_product / (ba * bc)
    return math.degrees(math.acos(cos))
a = tuple(map(float,input().split()))
b = tuple(map(float,input().split()))
c = tuple(map(float,input().split()))
print(angle(a,b,c))
```

#### W5A29:
Viết hàm trả về giá trị lớn nhất và nhỏ nhất của ba số a,b,c dưới dạng list [max, min].

```python
def min_max(a,b,c):
    g = [a,b,c]
    min = g[0]
    max = g[0]
    for i in g:
        if i < min:
            min = i
        if i > max:
            max = i
    return [max,min]
a,b,c = map(int,input().split())
print(min_max(a,b,c))
```

#### W5A30:
Viết hàm tính hệ số nhị thức C(n, k).

```python
def nhi_phuc(n,k):
    def giai_thua(x):
        a = 1
        for i in range(1,x+1):
            a *= i
        return a
    return giai_thua(n) / (giai_thua(k) * giai_thua(n-k))
n,k = map(int,input().split())
print(nhi_phuc(n,k))
```

#### W5A34:
Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau. Nếu n ≤ 0 trả về 0.

```python
def sum_1_to_n(n):
    a = 0
    if n <= 0:
        return 0
    for i in range(1,n+1):
        if len(str(i)) == len(set(str(i))):
            a += i
    return a
n = int(input())
print(sum_1_to_n(n))
```
