#### W5A3
Viết hàm kiểm tra một số có phải là số nguyên tố hay không?
def is_prime(n: int) -> bool:

```python
def is_prime(n: int) -> bool:
    for i in range(2,int(n**0.5)+1):
        if n%i == 0:
            return False
    else:
        return True
n = int(input())
print(is_prime(n))
```

#### W5A4
Viết hàm kiểm tra một số có phải số hoàn hảo hay không, xử lý trên dữ liệu kiểu số nguyên.
def is_perfect_number(n: int) -> bool:

```python
def is_perfect_number(n: int) -> bool:
    tong = 0
    for i in range(1,int(n**0.5)+1):
        if i**2 == n:
            tong += n
        elif i == 1:
            tong += 1
        
        else:
            if n%i == 0:
                tong += i+n/i
    if n == tong:
        return True
    else:
        return False

n = int(input())
print(is_perfect_number(n))
```

def find_first_index(lst: list, k: int) -> int:

```python
def find_first_index(lst: list, k: int) -> int:
    for i in range(len(lst)):
        if lst[i] == k:
            return i
    else:
        return -1

a = list(map(int,input().split()))
b = int(input())
print(find_first_index(a,b)+1)
```

#### W5A6
Viết hàm tính giai thừa. Nhập vào số nguyên n, trả về giai thừa của n (n!).
def factorial(n: int) -> int:

```python
def factorial(n: int) -> int:
    a = 1
    for i in range(1,n+1):
        a *= i
    return a

n = int(input())
print(factorial(n))
```

#### W5A7
Viết hàm giả lập một máy tính bỏ túi. Hàm thực hiện tính toán kết quả của phép toán giữa số num1 và num2 với toán tử operator (+, -, *, /). Kết quả làm tròn đến chữ số thập phân thứ hai.
def calculator(num1: float, num2: float, operator: str) -> float:

```python
def calculator(num1: float, num2: float, operator: str) -> float:
    if operator == "+":
        a = (num1+num2)
    elif operator == "-":
        a = (num1-num2)
    elif operator == "*":
        a = (num1*num2)
    elif operator == "/":
        a = (num1/num2)
    return round(a,2)

num1 = float(input())
num2 = float(input())
operator = input()
print(calculator(num1,num2,operator))
```

#### W5A8
Cho hai số nguyên x và y, hãy viết hàm trả về khoảng cách Hamming giữa hai số này.
def hamming_distance(x: int, y: int) -> int:

```python
def hamming_distance(x: int, y: int) -> int:
    a = x ^ y
    count = 0
    while a != 0:
        if a%2 == 1:
            count += 1
        a //= 2
    return count
x = int(input())
y = int(input())
print(hamming_distance(x,y))


```

#### W5A9
Viết hàm trả về tổng các chữ số của số nguyên dương n.
def sum_of_digits(n: int) -> int:

```python
def sum_of_digits(n: int) -> int:
    a = 0
    while n != 0:
        a += n%10
        n //= 10
    return a
n = int(input())
print(sum_of_digits(n))
```

#### W5A10
Viết hàm trả về True nếu hai từ a và b là đẳng cấu (isomorphic), ngược lại False. Giả sử hai từ đầu vào có cùng độ dài.
def are_isomorphic(a: str, b: str) -> bool:

```python
def are_isomorphic(a: str, b: str) -> bool:
    if len(a) != len(b):
        return False
    dict_a = {}
    dict_b = {}
    for i in range(len(a)):
        if a[i] not in dict_a:
            dict_a[a[i]] = b[i]
        elif dict_a[a[i]] != b[i]:
            return False
        if b[i] not in dict_b:
            dict_b[b[i]] = a[i]
        elif dict_b[b[i]] != a[i]:
            return False
    return True
a = input()
b = input()
print(are_isomorphic(a,b))
```

#### W5A12
Viết hàm divide_with_remainder(dividend, divisor) trả về bộ (q, r) sao cho dividend = divisor * q + r và 0 <= |r| < |divisor|.
def divide_with_remainder(dividend: int, divisor: int) -> tuple:

```python
def divide_with_remainder(dividend: int, divisor: int) -> tuple:
    if divisor != 0:
        q = dividend // divisor
        r = dividend % divisor
        if divisor > r > 0:
            return (q,r)
dividend = int(input())
divisor = int(input())
print(divide_with_remainder(dividend,divisor))
```

#### W5A15:
 Viết hàm is_right_triangle(a, b, c) trả về True nếu ba số có thể là ba cạnh của tam giác vuông.

```python
def is_right_triangle(a, b, c):
    if a**2 + b**2 == c**2 or a**2 + c**2 == b**2 or b**2 + c**2 == a**2:
        return True
    else:
        return False
a = int(input())
b = int(input())
c = int(input())
print(is_right_triangle(a, b, c))
```
