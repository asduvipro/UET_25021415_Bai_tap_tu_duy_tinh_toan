#### Bài 3 :
 Tính tam giác: Nhâp vào basố a,b,c bất kì. Kiểm tra xem nó có thể là
độ dài ba cạnh hay không, nếu không thì in ra màn hình 'Khong la ba canh
cua mot tam giac'. Ngượclại, kiểm tra xem nó là tam giác gì: Đều, Cân,
Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giácvà in ra màn
hình

```python
a,b,c = map(float,input().split())
if a + b > c and a + c > b and b + c > a:
    if a == b == c:
        print("Đây là tam giác đều")
    elif a == b or a == c or b == c:
        print("Đây là tam giác cân")
    elif a**2 + b**2 == c**2 or a**2 + c**2 == b**2 or b**2 + c**2 == a**2:
        print("Đây là tam giác vuông")
    else:
        print("Đây là tam giác thường")
else:
    print("Đây không phải là tam giác")
e = (a + b + c) / 2
dien_tich = (e * (e - a) * (e - b) * (e - c)) ** 0.5
chu_vi = a + b + c
print(f"diện tich của tam giác là: {dien_tich},chu vi của tam giác là: {chu_vi}")
```

#### Bài 5:
 Viết chương trình giải phương trình bậc 2 (ax2 + bx + c = 0)

```python
a,b,c = map(float,input().split())
t = b ** 2 - 4 * a * c
if a == 0:
    print("Phương trình này không phải phương trình bậc 2")
elif t == 0:
    print(f"Phương trình có nghiệm kép là: {-b / (2 * a)}")
elif t > 0:
    x1 = (-b + t ** 0.5) / (2 * a)
    x2 = (-b - t ** 0.5) / (2 * a)
    print(f"Phương trình có hai nghiệm phân biệt là: x1 = {x1}, x2 = {x2}")
else:
    print("Phương trình vô nghiệm")
```

#### Bài 9:
 Nhập vào thời gian 1 công việc nàođó là x giây. Hãy chuyểnđổi và viết
ra màn hình số thời gian trên dưới dạng bằng bao nhiêu giờ, bao nhiêu phút, bao
nhiêu giây.

```python
x = float(input())
so_gio = int(x // 3600)
so_phut = int((x % 3600) // 60)
so_giay = int(x % 60)
print(f"Bạn đã nhập {so_gio} giờ, {so_phut} phút, {so_giay} giây")
```
