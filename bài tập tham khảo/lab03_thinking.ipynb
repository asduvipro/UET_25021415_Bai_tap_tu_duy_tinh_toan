### Problem 1

#### Write a function that returns n raised to the power of 6.

```python
def power6(n):
    n = n**6
    return n
```

```python
assert power6(0) == 0
assert power6(1) == 1
assert power6(2) == 64
assert power6(-2) == 64
assert power6(10) == 1000000
```

### Problem 8

#### Write a function that returns the Euclidean distance between two points A(x1, y1) and B(x2, y2).

```python
def distance_ab(x1, y1, x2, y2):
    return ((x2 - x1)**2 + (y2 - y1)**2)**0.5
```

### Problem 9

#### Write a function that returns the angle ∠ABC in degrees given coordinates of points A(x1,y1), B(x2,y2), C(x3,y3).

```python
import math
def angle_ABC(x1, y1, x2, y2, x3, y3):
    a = (x1 - x2, y1 - y2)
    b = (x3 - x2, y3 - y2)
    c = (a[0] * b[0] + a[1] * b[1]) / ((a[0] ** 2 + a[1] ** 2) ** 0.5 * (b[0] ** 2 + b[1] ** 2) ** 0.5)
    return math.degrees(math.acos(c))
```

```python
assert round(angle_ABC(0, 0, 0, 1, 1, 1), 5) == 90.00000
assert round(angle_ABC(0, 0, 1, 0, 2, 0), 5) == 180.00000
assert round(angle_ABC(1, 0, 0, 0, 0, 1), 5) == 90.00000
assert round(angle_ABC(0, 0, 1, 1, 2, 2), 5) == 180.00000
assert round(angle_ABC(1, 0, 0, 0, -1, 0), 5) == 180.00000
```
