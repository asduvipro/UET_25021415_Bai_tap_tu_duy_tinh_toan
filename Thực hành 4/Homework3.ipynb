#### Bài 4:
 Viết chương trình in ra hình tam giác cân có độ cao h được nhập từ bàn phím

```python
h = int(input())
for i in range(1,h+1):
    if i == 1:
        print(" "*(h-1)+"*")
    elif i == h:
        print("*"*(2*h-1))
    else:
        print(" "*(h-i)+"*"+" "*(2*i-3)+"*")
```

#### Bài 9:
Nhập vào số nguyên dương a, kiểm tra xem a có phải là số chính phương hay không? Số chính phương là số có căn bậc hai là một số nguyên. VD: 1, 4, 9, 16, 25...

#### Bài 10:
Nhập vào số nguyên dương a, kiểm tra xem a có phải là số hoàn hảo hay không. Số hoàn hảo là số có tổng các ước (ước nhỏ hơn nó) bằng chính nó. VD: số 6, 28...

```python
a = int(input())
c = 0
for i in range(1,a):
    if a%i == 0:
        c = c+i
if c == a:
    print(f"{a} là một số hoàn hảo")
else:
    print(f"{a} không phải là một số hoàn hảo")
```

#### Bài 12:
Nhập số nguyên dương n, Cho $S(k) = 1 + 2 + 3 + \dots + k$. Tìm k sao cho $S(k)$ lớn nhất nhưng nhỏ hơn n.

```python
a = int(input())
c = 0
d = 0
while c+(d+1) < a:
    d += 1
    c = c+d
print(d)
```

#### Bài 13:
Nhập vào số nguyên dương A, tìm n nhỏ nhất sao cho:
$1 + 1/2 + 1/3 + 1/4 + \dots + 1/n > A$

```python
a = int(input())
c = 0
d = 0
while True:
    d += 1
    c = c + 1/d
    if c > a:
        break
print(d)
```
